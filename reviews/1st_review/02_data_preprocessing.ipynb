# Data Preprocessing

This notebook handles the initial data preprocessing of the raw dataset following these specific steps:
1. Load dataset
2. Remove duplicates
3. Handle missing values
4. Fix data types

```python
import pandas as pd
import numpy as np
import os
```

## 1. Load Dataset

```python
input_file = '../../data/coremof.xlsx'

print(f"Loading raw dataset from {input_file}...")
df = pd.read_excel(input_file, engine='openpyxl')
print(f"\nInitial Shape: {df.shape}")
```

## 2. Remove Duplicates

```python
initial_rows = len(df)
df = df.drop_duplicates()
print(f"Dropped {initial_rows - len(df)} duplicate rows. New Shape: {df.shape}")
```

## 3. Handle Missing Values

We fill numeric missing values with their mean and drop any remaining incomplete rows (e.g. for categorical columns).

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

df = df.dropna()
print(f"Shape after handling missing values: {df.shape}")
```

## 4. Fix Data Types

We convert object columns to numeric or datetime where appropriate.

```python
for col in df.columns:
    if df[col].dtype == 'object':
        try:
            df[col] = pd.to_numeric(df[col])
        except (ValueError, TypeError):
            try:
                df[col] = pd.to_datetime(df[col])
            except (ValueError, TypeError):
                pass

print("Data types after fixing:")
print(df.dtypes)
```

## Save Cleaned Data

```python
output_file = '../../data/coremof_cleaned.csv'
os.makedirs(os.path.dirname(output_file), exist_ok=True)
df.to_csv(output_file, index=False)
print(f"Cleaned dataset saved to {output_file}")
```
